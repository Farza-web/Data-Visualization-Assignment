```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic = pd.read_csv('/kaggle/input/datasets/lovishbansal123/titanic-dataset/Titanic-Dataset.csv')
iris = pd.read_csv('/kaggle/input/datasets/himanshunakrani/iris-dataset/iris.csv')
tips = pd.read_csv('/kaggle/input/datasets/saurabhbadole/restaurant-tips-dataset/tips.csv')
```

# Visualize at least 1 graph of each category. You can take help from the reference notebook

# Univariate Analysis

# 1. Pie chart

# Pie Chart for Titanic Dataset

```python
class_count = titanic['Pclass'].value_counts().sort_index()
plt.pie(class_count,labels=['1st Class', '2nd Class', '3rd Class'],autopct='%0.2f%%')
plt.title('Passenger Class Distribution')
plt.show()
```

# Pie Chart for Iris Dataset

```python
species_count = iris['species'].value_counts()
plt.pie(species_count,labels=species_count.index,autopct='%0.2f%%')
plt.title('Species Class Distribution')
plt.show()
```

# Pie Chart Tips Dataset

```python
day_count = tips['day'].value_counts()
plt.pie(day_count,labels=day_count.index,autopct='%0.2f%%')
plt.title('Day Class Distribution')
plt.show()
```

# 2. Count plot

# Titanic Count Plot

```python
sns.countplot(data = titanic,x='Pclass')
plt.title('Titanic Passenger Class Count Plot')
plt.show()
```

# Iris Dataset Count Plot

```python
sns.countplot(data=iris, x='species')
plt.title('Iris Species Count Plot')
plt.show()
```

```python
sns.countplot(data=tips, x='day')
plt.title('Tips Day Count Count Plot')
plt.show()
```

# 3. Kde + Rug plot

# Titanic Dataset Kde+Rug plot

```python
sns.kdeplot(data=titanic, x='Age')
sns.rugplot(data=titanic, x='Age')
plt.title('Titanic Age Distribution')
plt.show()
```

# Tips Dataset Kde+Rug plot

```python
sns.kdeplot(data=tips, x='total_bill')
sns.rugplot(data=tips, x='total_bill')
plt.title('Total Bill Distribution')
plt.show()
```

 # Tips Dataset Kde+Rug plot

```python
sns.kdeplot(data=iris, x='sepal_length')
sns.rugplot(data=iris, x='sepal_length')
plt.title('Iris Sepal Length Count Plot')
plt.show()
```

# Bivariate Analysis

# 1. Scatter plot

```python
sns.scatterplot(data=tips, x='total_bill', y='tip')
plt.title('Total Bill vs Tip')
plt.show()
```

```python
#Titanic Dataset 
sns.scatterplot(data=titanic, x='Age', y='Fare')
plt.title('Titanic - Age vs Fare')
plt.show()
```

```python
#Tips Dataset 
sns.scatterplot(data=tips,x='total_bill',y='tip')
plt.title('Tips - Bill vs Tip')
plt.show()
```

# 2. Line plot

```python
avg_tip = tips.groupby('day')['tip'].mean()
plt.plot(avg_tip.index, avg_tip.values, marker='o')
plt.title('Average Tip by Day')
plt.xlabel('Day')
plt.ylabel('Average Tip')
plt.show()
```

# 3. Joint plot

```python
#Iris Dataset
sns.jointplot(data=iris, x='sepal_length', y='sepal_width')
plt.show()
```

```python
# Titanic
sns.jointplot(data=titanic, x='Age',y='Fare')
plt.show()
```

# 4. Hexbin plot

# Multivariate Analysis

# 1. Scatter plot

```python
sns.scatterplot(data=tips, x='total_bill',y='tip',hue='sex',size='size')
plt.title('Total Bill vs Tip by Sex and Size')
plt.show()
```

# 2. Pair plot

```python
sns.pairplot(iris, hue='species')
plt.show()
```

# 3. Heatmap

```python
sns.heatmap(titanic.select_dtypes(include='number').corr(),
            annot=True)

plt.title('Titanic Correlation Heatmap')
plt.show()
```

# 4. FacetGrid plot

```python
h = sns.FacetGrid(tips, col='day', row='time')
h.map(sns.histplot, 'total_bill')
plt.show()
```

# 5. PairGrid

```python
g = sns.PairGrid(iris, hue='species')
g.map_diag(sns.histplot)
g.map_upper(sns.scatterplot)
g.map_lower(sns.kdeplot)
plt.show()
```

# 6. Joint plot

```python
sns.jointplot(data=tips, x='total_bill',y='tip',hue='sex')
plt.show()
```

# 7. Boxen plot

```python
sns.jointplot(data=tips,x='total_bill',  y='tip', hue='sex')
plt.show()
```
